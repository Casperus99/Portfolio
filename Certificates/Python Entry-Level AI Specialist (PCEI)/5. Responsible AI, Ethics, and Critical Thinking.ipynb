# **Section 5**: Responsible AI, Ethics, and Critical Thinking

Candidates should understand ethical and safety considerations in AI systems, including bias, fairness, privacy, and misuse. They should be able to analyze the societal and economic impact of AI, explain principles of responsible and human-centered AI, and critically evaluate AI-generated outputs for accuracy, consistency, and trustworthiness.

## 5.1 – Identify Ethical Risks and Safety Concerns in AI

Keywords: bias, discrimination, fairness, privacy, safety, hallucinations, harmful outputs, responsible AI

### A. Identify common ethical risks such as bias, discrimination, unfair outcomes, and harmful stereotypes

### B. Recognize privacy concerns related to data collection, use, and storage

### C. Describe safety issues such as hallucinations, misinformation, and inappropriate model outputs

### D. Distinguish between safe and unsafe use of AI technologies in everyday scenarios

## 5.2 - Apply Basic Safety and Security Practices When Using AI Systems

Keywords: AI safety, AI security, sensitive data, data protection, misuse prevention, safe interaction, responsible use, escalation

### A. Recognize which types of information should not be shared with AI systems (e.g., personal data, passwords, private documents)

### B. Identify unsafe or risky system interactions such as attempts to bypass controls, manipulate model behavior, or access restricted content

### C. Follow basic security and safety guidelines when using AI tools, including data minimization, responsible input handling, and adherence to terms of use

### D. Understand when to escalate or stop using an AI system if unexpected or potentially harmful behavior occurs

## 5.3 - Analyze the Social and Economic Impact of AI

Keywords: societal impact, workforce, automation, reskilling, digital divide, economic impact, transformation

### A. Describe how AI affects society, education, industry, and public services

### B. Identify ways AI transforms jobs and creates a need for reskilling

### C. Recognize the risks of job displacement and widening inequalities (digital divide)

### D. Evaluate both positive and negative impacts of AI adoption

## 5.4 - Explain Principles of Responsible, Transparent, and Human-Centered AI

Keywords: responsible AI, transparency, accountability, explainability, human-in-the-loop, oversight, verification

### A. Explain the principles of responsible AI (fairness, transparency, accountability, explainability)

### B. Describe the concept of “human-in-the-loop” and why human oversight is required

### C. Recognize the importance of verifying AI outputs, checking sources, and applying human judgment

### D. Identify practices that promote safe and trustworthy AI deployment

## 5.5 - Apply Critical Thinking to Evaluate AI Outputs

Keywords: critical thinking, verification, hallucinations, unsupported claims, contradictions, reasoning, human judgment

### A. Identify when AI-generated responses are incorrect, misleading, or inconsistent

### B. Detect hallucinations, unsupported claims, and logical errors in AI outputs

### C. Cross-check AI answers using reasoning and external knowledge

### D. Determine when human decision-making should override automated outputs
