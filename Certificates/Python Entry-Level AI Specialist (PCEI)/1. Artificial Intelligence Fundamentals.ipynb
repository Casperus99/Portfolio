# **Section 1**: Artificial Intelligence Fundamentals

Candidates should be able to define Artificial Intelligence (AI), describe its fundamental components, and distinguish between core terms used in AI systems. They should understand the difference between narrow and general AI, explain how AI agents interact with their environment, and identify common real-world applications of AI. 

## 1.1 – Define Artificial Intelligence and Its Core Concepts

Keywords: AI, agent, environment, inference, narrow AI, general AI, data, algorithm, conceptual model, input-processing-output, real-world applications 

### A. Define Artificial Intelligence and related terminology  (AI, agent, environment, inference)

- **Artificial Intelligence (AI):** The simulation of human intelligence processes by machines (especially computer systems). It is the capability of a machine to imitate intelligent human behavior, such as learning, reasoning, and problem-solving.
- **Agent:** An entity (software program or robot) that perceives its environment through sensors and takes actions upon that environment through actuators to achieve specific goals.
- **Environment:** Everything outside the agent. It is the domain or context in which the AI agent operates, observes, and acts.
- **Inference:** The process of drawing conclusions or making predictions based on available data and learned patterns. In Machine Learning, it is the phase where a trained model is used to predict outputs for new, unseen data.

### B. Distinguish between Narrow AI and General AI

*   **Narrow AI (Weak AI):** 
    *   Designed and trained for a **specific, limited task** (e.g., playing chess, translating text, facial recognition).
    *   Cannot perform tasks outside its predefined scope.
*   **General AI (Strong AI / AGI - Artificial General Intelligence):** 
    *   A theoretical form of AI that possesses **human-level cognitive abilities**.
    *   Could understand, learn, and apply knowledge across any generalized task, just like a human. 

### C. Identify the components of an AI system

*   **Data:** The raw information (numbers, text, images, video) fed into the system. It is the "fuel" used to train algorithms and make predictions.
*   **Algorithms:** A set of mathematical rules, instructions, or statistical procedures that the computer follows to learn patterns from data or solve problems.
*   **Conceptual Models:** The mathematical representation created after an algorithm has been trained on data. The model holds the "learned" rules and is used to make decisions.
*   **The Workflow (Input → Processing → Output):**
    *   **Input:** The system receives raw data from the environment (e.g., text from a user, an image from a camera).
    *   **Processing:** The AI applies its algorithm/trained model to analyze the input data and perform inference.
    *   **Output:** The system produces a result, prediction, or action (e.g., a text response, categorizing an image as a "cat").

### D. Recognize examples of AI in everyday life

*   **Search:** Google Search using AI to understand query intent, rank web pages, and provide auto-complete suggestions.
*   **Recommendations:** Netflix, YouTube, or Spotify algorithms suggesting movies, videos, or songs based on user viewing/listening history.
*   **Assistants:** Virtual voice assistants like Siri, Alexa, or Google Assistant using Natural Language Processing (NLP) to understand and execute voice commands.
*   **Navigation:** Google Maps or Waze predicting traffic patterns and finding optimal routes in real-time.
*   **Image Processing:** Smartphone FaceID (facial recognition), automated photo categorization, or medical AI detecting anomalies in X-rays.
*   **Text Processing:** Email spam filters, grammar checkers (like Grammarly), machine translation (Google Translate), and conversational chatbots (like ChatGPT).

## 1.2 - Distinguish Major Subfields of AI

Keywords: machine learning, deep learning, NLP, computer vision, robotics, generative AI, LLMs 

### A. Describe the main subfields of AI (Machine Learning, Deep Learning, NLP, Computer Vision, Robotics)

*   **Machine Learning (ML):** A subset of AI where algorithms learn patterns from data and make decisions or predictions without being explicitly programmed for that specific task.
*   **Deep Learning (DL):** A specialized subset of ML that uses multi-layered artificial neural networks (hence "deep") to process highly complex and unstructured data, like images or audio.
*   **NLP (Natural Language Processing):** The branch of AI focused on the interaction between computers and human language. It enables machines to understand, interpret, and generate text or speech.
*   **Computer Vision (CV):** The field that allows computers to "see". It enables machines to interpret, analyze, and extract meaningful information from visual data (digital images and videos).
*   **Robotics:** A multidisciplinary field combining AI, engineering, and sensors to build physical or virtual agents (robots) that can execute tasks and interact with the physical environment.

### B. Match problems to appropriate subfields (e.g., text summarization → NLP)

*   *Text summarization, language translation, sentiment analysis, spam filtering* → **NLP**
*   *Facial recognition, medical scan analysis, object detection in self-driving cars* → **Computer Vision**
*   *Predicting house prices, customer segmentation, recommendation systems* → **Machine Learning**
*   *Automated assembly line arms, autonomous drones navigating a warehouse* → **Robotics**
*   *Generating a realistic image of a cat that doesn't exist* → **Deep Learning / Generative AI / Computer Vision**

### C. Describe how generative AI relates to these subfields

*   **Generative AI** is a highly advanced subset of **Deep Learning** (and therefore **Machine Learning**).
*   Unlike traditional ML which typically classifies data or predicts outcomes (e.g., "is this a cat?"), Generative AI *creates entirely new, original content* (text, images, code, audio) based on learned patterns.
*   **Relation to NLP:** When Generative AI focuses on text, it uses **LLMs (Large Language Models)** like ChatGPT. LLMs sit perfectly at the intersection of Generative AI and NLP.
*   **Relation to Computer Vision:** Generative AI tools (like DALL-E or Midjourney) apply Computer Vision principles in reverse to generate new visual content from text prompts.

## 1.3 - Describe How AI Systems Learn From Data

Keywords: training, inference, features, labels, predictions, labeled data, unlabeled data, feedback, data quality, learning from data 

### A. Explain the idea of learning from data using simple examples

*   **Concept:** Instead of being explicitly programmed with strict rules (e.g., "IF it has 4 wheels and an engine, THEN it is a car"), AI systems find patterns on their own by analyzing large amounts of examples.
*   **Example:** To teach an AI to recognize spam, you don't write a rule for every possible bad word. You feed the system 10,000 emails marked as "spam" and 10,000 marked as "safe". The AI mathematically discovers the patterns (frequent words, sender types) that distinguish the two.

### B. Distinguish training from inference

*   **Training:** The *learning phase*. It is the computationally heavy process of feeding historical data into an algorithm so it can find patterns, adjust its internal parameters, and build a conceptual model. 
*   **Inference:** The *applying phase*. It is the process of using the already-trained model to make **predictions** or decisions on new, previously unseen data.

### C. Describe features, labels, and how simple models use them

*   **Features:** The input variables, characteristics, or attributes of the data used to make a prediction. *(e.g., To predict a house price, features would be: square footage, number of bedrooms, age of the house).*
*   **Labels:** The target output, correct answer, or category that the model is trying to predict. *(e.g., The actual price of the house).*
*   **How models use them:** During training, the model analyzes the **features** alongside their corresponding **labels** to learn the mathematical relationship between them. Later (during inference), it takes new features to guess the missing label.

### D. Recognize the difference between labeled and unlabeled data in basic learning scenarios

*   **Labeled Data:** Data that has a known answer or tag attached to it (e.g., a photo with a "Cat" tag, or an email marked as "Spam"). It is used in *Supervised Learning* to teach the model by example.
*   **Unlabeled Data:** Data without any tags or correct answers attached (e.g., thousands of articles without categories). The model must discover hidden structures or groupings on its own. It is used in *Unsupervised Learning*.

### E. Recognize feedback loops in AI systems

*   **Feedback:** A continuous improvement mechanism where the real-world outcome of an AI's prediction is collected and fed back into the system as new training data.
*   **Example:** A recommendation engine suggests a video. If you click "Not interested" (feedback), the system updates its dataset and adjusts the model to make better recommendations in the future.

### F. Identify how data quality affects learning (noise, bias, gaps)

*   **General Rule:** "Garbage in, garbage out". The performance of an AI model relies entirely on **data quality**. Poor data leads to poor learning.
*   **Noise:** Random errors, corrupted data, or irrelevant information within the dataset that confuses the algorithm and degrades accuracy.
*   **Bias:** A systematic error where the dataset is skewed and does not accurately represent the real world, leading the AI to make unfair or prejudiced predictions.
*   **Gaps:** Missing information or unrepresented groups in the dataset. If a model never sees a specific scenario during training, it will fail to handle it during inference.

## 1.4 - Identify the Capabilities and Limitations of AI Systems

Keywords: capabilities, limitations, hallucinations, generalization, robustness, context, failure cases 

### A. Identify typical capabilities of AI systems (classification, prediction, pattern recognition)

*   **Classification:** The ability to assign data into predefined categories based on learned features (e.g., sorting emails into "Spam" or "Inbox", identifying a tumor as "Malignant" or "Benign").
*   **Prediction (Regression/Forecasting):** The ability to estimate future continuous values or trends based on historical data (e.g., predicting tomorrow's temperature, stock market prices, or customer lifetime value).
*   **Pattern Recognition:** Identifying recurring structures, trends, or anomalies in massive datasets that would be impossible for humans to spot (e.g., detecting fraudulent credit card transactions, clustering similar users).

### B. Recognize common limitations (lack of context, hallucinations, misclassification)

*   **Lack of Context:** AI models do not possess human common sense, empathy, or true understanding of the real world. They operate purely on mathematical probability and can easily fail to grasp nuance, sarcasm, or broader situational context.
*   **Hallucinations:** A specific limitation of Generative AI (like LLMs). The model confidently produces false, fabricated, or nonsensical information because it is designed to predict the next plausible word, not to fact-check against reality.
*   **Misclassification:** When a model incorrectly categorizes an input (e.g., mistaking a stop sign with a sticker on it for a speed limit sign). This often highlights a lack of **robustness** (the system's ability to cope with noise, errors, or adversarial inputs). 
*   **Failure Cases:** Situations where the AI completely breaks down or produces harmful outputs because it encountered a scenario entirely outside its training scope.

### C. Explain why AI performance depends on data quality and diversity

*   **Quality over Quantity:** A model trained on a small, clean, and accurate dataset will often outperform a model trained on a massive, noisy dataset. 
*   **Diversity & Generalization:** Data must be diverse to represent all possible real-world scenarios. Good diversity leads to **generalization** — the ability of an AI model to perform accurately on new, unseen data. If data lacks diversity (e.g., facial recognition trained only on light-skinned faces), the model will fail on underrepresented groups (bias).

### D. Distinguish situations where AI is appropriate vs. inappropriate

*   **Appropriate:** 
    *   Tasks involving analyzing huge volumes of data quickly.
    *   Repetitive, pattern-based tasks (automation).
    *   Scenarios where minor errors are acceptable and easily corrected (e.g., movie recommendations, drafting an email).
*   **Inappropriate:** 
    *   Tasks requiring deep empathy, moral judgment, or complex ethical decision-making (e.g., final sentencing in a court of law).
    *   Situations with zero tolerance for error without human oversight (e.g., fully autonomous weapons, critical life-or-death medical diagnoses).
    *   Problems where no historical data exists to train the model.

## 1.5 - Identify Requirements for Building AI-Driven Solutions

Keywords: AI requirements, success metrics, problem definition, stakeholders, evaluation, project goals, AI adoption

### A. Identify requirements needed to implement an AI solution (data availability, problem definition, objectives)

Before writing any code or training a model, specific foundational requirements must be met:
*   **Problem Definition:** Clearly understanding and defining the specific problem to be solved. AI should not be "a solution looking for a problem." You must know exactly *what* you are trying to predict, classify, or automate.
*   **Data Availability:** AI relies on data. You must ensure that you have access to a sufficient volume of relevant, high-quality, and legally compliant data to train the model. If data is unavailable or heavily restricted, the AI project cannot proceed.
*   **Objectives:** Setting clear, measurable goals (both technical and business-oriented). What does success look like? (e.g., "Reduce manual data entry time by 50%" or "Achieve 95% accuracy in spam detection").

### B. Recognize success metrics for evaluating AI systems

To know if an AI system is successful, you must measure its performance using specific metrics:
*   **Technical Metrics (Classification):**
    *   **Accuracy:** The overall percentage of correct predictions.
    *   **Precision:** Out of all positive predictions the model made, how many were actually correct? (Important when false positives are costly).
    *   **Recall (Sensitivity):** Out of all actual positive cases, how many did the model find? (Important when false negatives are dangerous, e.g., medical diagnosis).
    *   **F1-Score:** The balance (harmonic mean) between precision and recall.
*   **Technical Metrics (Regression):**
    *   **Mean Absolute Error (MAE) / Mean Squared Error (MSE):** Measures how far off the model's numerical predictions are from the actual values.
*   **Business Metrics:** Return on Investment (ROI), processing time saved, customer satisfaction score, or cost reduction.

### C. Distinguish problems suitable for AI from those unsuitable

*   **Suitable for AI:**
    *   Tasks that are highly repetitive and operate on large volumes of data.
    *   Problems where the rules are too complex to write manually using `if/else` statements, but patterns can be learned from examples (e.g., recognizing faces, translating languages).
    *   Environments where continuous learning and adaptation to new data are beneficial.
*   **Unsuitable for AI:**
    *   Problems with very little to no historical data available for training.
    *   Simple tasks that can be perfectly solved with traditional, strict rules (where a basic script is cheaper, 100% predictable, and sufficient).
    *   Situations requiring absolute "common sense," genuine human empathy, or complex moral/ethical judgments.
    *   Highly critical systems where a "black box" model (unexplainable decision-making) is legally or ethically unacceptable.
