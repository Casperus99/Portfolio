# **Section 6**: AI Projects, Collaboration, and Communication

Candidates should be able to identify opportunities for AI solutions, plan and implement small AI-driven tasks using Python, and evaluate their results. They should demonstrate effective collaboration practices and communicate findings clearly to technical and non-technical audiences, using appropriate visualizations and explanations.

## 6.1 – Identify Opportunities for AI Solutions in Real-World Contexts

Keywords: AI opportunities, problem identification, project goals, constraints, feasibility, data availability

### A. Identify simple, realistic problems that can be supported by AI (classification, prediction, summarization, recommendation)

### B. Distinguish between problems that are appropriate for AI and those that require traditional programming or human judgment

### C. Recognize project goals, inputs, outputs, and basic constraints when planning an AI solution

### D. Describe how data availability, quality, and domain knowledge influence project feasibility

## 6.2 - Recognize Basic Cost Considerations When Planning AI Projects

Keywords: AI cost, budgeting, resource estimation, compute usage, API limits, feasibility, efficiency, cost–benefit consideration

### A. Identify common factors that influence the cost of using AI systems, such as data volume, model complexity, computational requirements, and use of external AI services

### B. Distinguish between free, low-cost, and paid AI tools or APIs, and understand how usage limits may affect project design

### C. Estimate whether a small AI project is feasible within typical resource constraints (time, storage, compute, or budget)

### D. Recognize situations where simpler approaches (rule-based logic, smaller models, offline tools) may reduce cost without affecting project goals

## 6.3 - Apply Basic AI Development Steps to a Small Project

Keywords: data preparation, simple models, Python implementation, evaluation, accuracy, project iteration

### A. Gather or load a small dataset suitable for a simple AI task

### B. Prepare and transform data (cleaning, selecting attributes, summarizing)

### C. Implement a simple solution using Python (rule-based classification, k-NN-style logic, basic predictive workflow)

### D. Evaluate the output of the solution using simple metrics (accuracy, counts, comparisons)

### E. Identify limitations and areas for improvement in the project

## 6.4 - Collaborate Effectively in AI Projects

Keywords: collaboration, teamwork, version control, code review, roles, documentation, reproducibility

### A. Describe the roles commonly involved in AI projects (analyst, developer, domain expert, reviewer)

### B. Use basic collaboration practices such as version control, shared responsibilities, and peer feedback

### C. Participate in pair programming, code reviews, or simple teamwork workflows

### D. Recognize the importance of documenting decisions and maintaining reproducibility

## 6.5 - Communicate AI Findings Clearly to Technical and Non-Technical Audiences

Keywords: communication, presentation, visualization, reporting, audience adaptation, summarization

### A. Present AI project results using charts, summaries, and simple explanations

### B. Explain model outputs, limitations, and errors in clear, accessible language

### C. Write short technical summaries describing methods, results, and conclusions

### D. Tailor communication style based on the target audience (e.g., classmates vs. non-technical stakeholders)
